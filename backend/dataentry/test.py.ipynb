```python
new_df=db[['id', 'title','published_date','authors', 'first_author','summary']]
```

```python
print(new_df)
```

```python
import numpy as np

for index, row in new_df.iterrows():
    if row['id'].startswith('cs-'):
        new_df.at[index, 'pdf_link'] = "https://arxiv.org/pdf/cs/" + row['id'][3:]
    else:
        new_df.at[index, 'pdf_link'] = np.nan
```

```python
for index, row in new_df.iterrows():
    # access data

    if row['id'].startswith('cs-'):
        print(row['id'],row['pdf_link'])
```

```python
new_df.head()
```

```python
final_df = new_df.loc[
    new_df['id'].str.startswith('cs-', na=False),
    ['id', 'title', 'published_date', 'summary','authors', 'first_author', 'pdf_link']
]
```

```python
print(final_df)
```

```python
final_df.to_csv("new_final_data.csv", index=False)
```
